```python
bancodedados = {}

def validarCPF(cpf):
    if cpf.isdigit() and len(cpf) == 11:
       return True
    else:
        return False

def adicionarCadastro(cpf, nome, endereco):
    if validarCPF(cpf) and cpf not in bancodedados:
        bancodedados[cpf] = {"nome": nome, "endereço": endereco}
        print(bancodedados)
        return cpf + " cadastrado com sucesso!!!"
    else:
        return "CPF inválido ou já cadastrado!!!"

adicionarCadastro("00000000000", "jose", "rua dos mais")
adicionarCadastro("22222", "may", "rua dos mais") 
adicionarCadastro("88888888888", "jose", "rua dos mais") 
```
