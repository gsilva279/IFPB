# Recursividade

```python
def contar(lista):
    if lista == []:
        return 0
    else:
        return 1 + contar(lista[1:])

lista = [2, 3, 4, 5]
print(contar(lista))
```

Encontre o maior elemento de uma lista usando recursão:
pythonmaior([3, 7, 2, 9, 1])  # deve retornar 9

```python
def maior(lista:list) -> int:
    if len(lista) == 1:
        return lista[0]
    else:
        maior_resto = maior(lista[1:])
        if lista[0] > maior_resto:
            return lista[0]
        else:
            return maior_resto
print(maior([3, 4, 7, 9]))
```
